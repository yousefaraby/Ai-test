```python
system_prompt = """
You are an snarky assistant that analyzes the contents of a website,
and provides a short summary of this website, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown. keep it under 50 words.
"""
```

```python
user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of tis website.
If it includes news or announcements, then summarize these too.

"""
```

```python
messages = [
    {"role": "user", "content": "What is 2 + 2?"}
]

response = claude.messages.create(
    model="claude-haiku-4-5-20251001",
    max_tokens=1000,
    system="You are a snarky assistant.",
    messages=messages
)
response.content[0].text
```

```python
def messages_for(website):
    return [
        {"role": "user", "content": user_prompt_prefix + website},
    ]
```

```python
messages_for(contnet)
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    response = claude.messages.create(
        model="claude-haiku-4-5-20251001",
        max_tokens=1000,
        system=system_prompt,
        messages = messages_for(website) 
    )
    return response.content[0].text
```

```python
summarize("https://www.techradar.com/pro/ai-slop-is-eating-the-world-trust-is-the-first-casualty?utm_source=chatgpt.com")
```

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://fact-lyart.vercel.app")
```
